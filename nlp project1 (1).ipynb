```python

text_column = "message"


df["original_text"] = df[text_column].copy()


df["cleaned_text"] = df["original_text"].astype("string").str.strip()


df["cleaned_text"] = df["cleaned_text"].str.replace(
    r"\s+", " ", regex=True
)


display(df[["original_text", "cleaned_text"]].head())
```

```python
import re
import pandas as pd

def clean_text(text):

    
    if pd.isna(text):
        return text

    text = str(text)

    
    text = re.sub(r"<[^>]+>", " ", text)

    
    text = re.sub(r"[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]", " ", text)

    
    text = re.sub(r"\s+", " ", text)

    return text.strip()



df["cleaned_text"] = df["cleaned_text"].apply(clean_text)


display(df[["original_text", "cleaned_text"]].head())
```

```python
import re
import pandas as pd

def remove_extra_whitespaces(text):

    if pd.isna(text):
        return text

    text = str(text)

    
    text = re.sub(r"\s+", " ", text)

    return text.strip()



df["cleaned_text"] = df["cleaned_text"].apply(
    remove_extra_whitespaces
)


display(df[["original_text", "cleaned_text"]].head())
```

```python

sample_text = df["message"].dropna().iloc[0]


cleaned_sample = clean_text(sample_text)


print("Original Text:")
print(sample_text)

print("\n" + "=" * 50)


print("Cleaned Text:")
print(cleaned_sample)
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

df["tokens"] = df["cleaned_text"].apply(
    lambda text: word_tokenize(text) if isinstance(text, str) else []
)

display(df[["cleaned_text", "tokens"]].head())
```

```python
for i in range(5):
    print(f"Email {i+1}")
    print("Original Cleaned Text:")
    print(df["cleaned_text"].iloc[i])

    print("\nTokens:")
    print(df["tokens"].iloc[i])

    print("\n" + "="*80 + "\n")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
def get_pos_tags(text):
    if not isinstance(text, str):
        return []

    doc = nlp(text)

    return [(token.text, token.pos_) for token in doc]

df["pos_tags_spacy"] = df["cleaned_text"].apply(get_pos_tags)
```

```python
for i in range(5):
    print(f"Email {i+1}")

    print("\nTokens:")
    print(df["tokens"].iloc[i])

    print("\nPOS Tags using spaCy:")
    print(df["pos_tags_spacy"].iloc[i])

    print("\n" + "="*80 + "\n")
```
